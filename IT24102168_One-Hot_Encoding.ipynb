```python
file_path = "/content/drive/MyDrive/2025-Y2-S1-MLB-B8G1-03/results/outputs/cleaned_student_depression_data.csv"
df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
```

```python
#display first 5 rows
df.head()
```

```python
# Manual one-hot encoding for Gender
df['Gender_Male'] = (df['Gender'] == 'Male').astype(int)
df['Gender_Female'] = (df['Gender'] == 'Female').astype(int)

print("DataFrame with manual one-hot encoding for Gender:")
df.head(5)
```

```python
# Drop the original gender column
df = df.drop('Gender', axis=1)
df.head()
```

```python
# Plot the distributions of the one-hot encoded Gender columns
diet_columns = ['Gender_Male', 'Gender_Female']
for col in diet_columns:
    plt.figure(figsize=(6,4))
    sns.countplot(x=col, data=df)
    plt.title(f"{col} Distribution")

save_path = f"/content/drive/MyDrive/2025-Y2-S1-MLB-B8G1-03/results/eda_visualizations/{col}_IT24102168_encoding_distributions_Gender.png"

# Save the plot
plt.savefig(save_path, bbox_inches="tight")
plt.close()   # Close the figure to avoid overlapping plots

print(f"Saved: {save_path}")
```

```python
#one-hot encoding for Dietary Habits
df['Diet_Healthy'] = (df['Dietary_Habits'] == 'Healthy').astype(int)
df['Diet_Moderate'] = (df['Dietary_Habits'] == 'Moderate').astype(int)
df['Diet_Unhealthy'] = (df['Dietary_Habits'] == 'Unhealthy').astype(int)

print("DataFrame with manual one-hot encoding:")
df.head(5)
```

```python
#Drop the original Dietary_Habits column
df = df.drop('Dietary_Habits', axis=1)
df.head()
```

```python
# Plot the distributions of the one-hot encoded Dietary_Habits columns
diet_columns = ['Diet_Healthy', 'Diet_Moderate', 'Diet_Unhealthy']
for col in diet_columns:
    plt.figure(figsize=(6,4))
    sns.countplot(x=col, data=df)
    plt.title(f"{col} Distribution")

save_path = f"/content/drive/MyDrive/2025-Y2-S1-MLB-B8G1-03/results/eda_visualizations/{col}_IT24102168_encoding_distributions_dietary_habits.png"

# Save the plot
plt.savefig(save_path, bbox_inches="tight")
plt.close()   # Close the figure to avoid overlapping plots

print(f"Saved: {save_path}")
plt.show()
```

```python
# Save processed dataset / features
# Define save path for processed dataset
save_path = "/content/drive/MyDrive/2025-Y2-S1-MLB-B8G1-03/results/outputs/student_depression_after_one_hot_encoding.csv"

# Save as CSV
df.to_csv(save_path, index=False)

print(f"Preprocessed dataset saved successfully at: {save_path}")
```
