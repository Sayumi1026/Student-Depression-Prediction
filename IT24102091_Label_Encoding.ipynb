```python
file_path = "/content/drive/MyDrive/2025-Y2-S1-MLB-B8G1-03/results/outputs/student_depression_after_one_hot_encoding.csv"
df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
```

```python
#display first 5 rows
df.head()
```

```python
# Apply preprocessing techniques
#Label encoding for Family History of Mental Illness and Suicidal Thoughts

# Family History of Mental Illness
# The column name in the dataframe is 'Family History of Mental Illness'
df['Family_History'] = df['Family_History'].map({'No': 0, 'Yes': 1})

# Suicidal Thoughts
# The column name in the dataframe is 'Have you ever had suicidal thoughts ?'
df['Suicidal_Thoughts'] = df['Suicidal_Thoughts'].map({'No': 0, 'Yes': 1})
```

```python
#display first 5 rows
df.head()
```

```python
#Family History distribution after encoding
plt.figure(figsize=(6,4))
sns.countplot(x="Family_History", data=df)
plt.title("Family History of Mental Illness Distribution After Encoding")

# Define save path
save_path = "/content/drive/MyDrive/2025-Y2-S1-MLB-B8G1-03/results/eda_visualizations/IT24102091_label_encoding_family_history_distribution.png"

# Save the plot
plt.savefig(save_path, bbox_inches="tight")

print(f"Saved: {save_path}")

plt.show()
```

```python
#Suicidal Thoughts distribution after encoding
sns.countplot(x="Suicidal_Thoughts", data=df)
plt.title("Suicidal Thoughts Distribution After Encoding")

# Define save path
save_path = "/content/drive/MyDrive/2025-Y2-S1-MLB-B8G1-03/results/eda_visualizations/IT24102091_label_encoding_Suicidal_Thoughts_Distribution.png"

# Save the plot
plt.savefig(save_path, bbox_inches="tight")

print(f"Saved: {save_path}")
plt.show()
```
